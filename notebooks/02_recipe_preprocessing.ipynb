```python
import pandas as pd
import sys
import os

# --- Début de la correction pour le chemin d'importation des modules 'src' ---
project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
if project_root not in sys.path:
    sys.path.append(project_root)
# --- Fin de la correction ---

# Importer les fonctions de notre module de prétraitement (on importe load_and_process_moroccan_recipes)
from src.preprocessing import process_raw_recipes_pipeline, load_and_process_moroccan_recipes, build_recipe_document
```

```python
import pandas as pd
import os
import re # Nécessaire pour les fonctions de nettoyage de texte si elles sont appelées
from IPython.display import display # Pour afficher les DataFrames dans un notebook

# --- Définition explicite du dossier de sortie ---
output_dir = '../data/processed'
os.makedirs(output_dir, exist_ok=True) # S'assurer que le dossier existe
# -------------------------------------------------

print(f"Dossier de sortie défini : {output_dir}")
print("\nDébut du nettoyage avancé et de la déduplication du fichier existant (préservant la colonne 'document')...")
```

```python
input_file_path = os.path.join(output_dir, 'all_recipes_combined_processed.csv')
try:
    # Charger avec toutes les colonnes par défaut, y compris 'document'
    df_recipes = pd.read_csv(input_file_path)
    print(f"Fichier '{input_file_path}' chargé avec succès. Taille initiale : {len(df_recipes)} recettes.")
except FileNotFoundError:
    print(f"Erreur : Le fichier '{input_file_path}' n'a pas été trouvé.")
    print("Assure-toi que tu as exécuté les étapes précédentes pour créer ce fichier.")
    exit() # Quitte si le fichier n'est pas trouvé
```

```python
print("\nApplication du nettoyage des noms et de la déduplication...")
print("\nRemplacement des NaN par des chaînes vides ('') dans 'ingredients_text' et 'steps_text'...")

df_recipes['ingredients_text'].fillna('', inplace=True)
df_recipes['steps_text'].fillna('', inplace=True)

print("Remplacement terminé. Les colonnes devraient maintenant être propres.")

# Vérification (optionnel, pour s'assurer qu'il n'y a plus de NaN)
print("\nVérification des valeurs NaN après remplacement :")
print(df_recipes[['ingredients_text', 'steps_text']].isna().sum())
# 2.1 Supprimer les recettes avec le nom exact "Nom de recette inconnu"
initial_rows = len(df_recipes)
df_filtered_by_name = df_recipes[df_recipes['name'] != 'Nom de recette inconnu'].copy()

print(f"Recettes après suppression des noms 'Nom de recette inconnu' : {len(df_filtered_by_name)} (supprimées: {initial_rows - len(df_filtered_by_name)})")
```

```python
print("\nRemplacement des NaN par des chaînes vides ('') dans 'ingredients_text' et 'steps_text'...")

df_recipes['ingredients_text'].fillna('', inplace=True)
df_recipes['steps_text'].fillna('', inplace=True)

print("Remplacement terminé. Les colonnes devraient maintenant être propres.")

# Vérification (optionnel, pour s'assurer qu'il n'y a plus de NaN)
print("\nVérification des valeurs NaN après remplacement :")
print(df_recipes[['ingredients_text', 'steps_text']].isna().sum())
```

```python
# 2.2 Identifier et garder les doublons avec le plus d'informations
# Créer une colonne pour évaluer la quantité d'informations
# Les colonnes sont maintenant garanties d'être des chaînes
df_filtered_by_name['info_length'] = df_filtered_by_name.apply(
    lambda row: len(row['ingredients_text']) + len(row['steps_text']), # safe_str() n'est plus strictement nécessaire ici car fillna a été fait avant
    axis=1
)

# Trier par nom, puis par longueur d'information décroissante
# Garder la première occurrence (celle avec le plus d'infos) pour chaque nom
df_cleaned_recipes = df_filtered_by_name.sort_values(
    by=['name', 'info_length'],
    ascending=[True, False]
).drop_duplicates(subset=['name'], keep='first')

# Supprimer la colonne temporaire 'info_length'
df_cleaned_recipes = df_cleaned_recipes.drop(columns=['info_length'])

print(f"Recettes après déduplication (gardant la plus complète) : {len(df_cleaned_recipes)} (doublons supprimés: {len(df_filtered_by_name) - len(df_cleaned_recipes)})")
print("Aperçu du DataFrame nettoyé et dédupliqué (avec colonne 'document' préservée) :")
display(df_cleaned_recipes.head())
```

```python
# 3. Sauvegarde du Dataset Nettoyé
output_file_path_cleaned = os.path.join(output_dir, 'all_recipes_combined_processed2.csv')
df_cleaned_recipes.to_csv(output_file_path_cleaned, index=False)

print(f"\nDataset nettoyé et dédupliqué sauvegardé sous : {output_file_path_cleaned}")
print(f"Taille finale du dataset : {len(df_cleaned_recipes)} recettes.")
```

```python
import pandas as pd
import os

# --- Définition explicite du dossier de sortie ---
output_dir = '../data/processed'
os.makedirs(output_dir, exist_ok=True)
# -------------------------------------------------

print(f"Dossier de sortie défini : {output_dir}")
print("\nDébut du nettoyage final : suppression des lignes sans ingrédients_text ou steps_text...")

# Nom du fichier à nettoyer (ton fichier déjà traité)
input_file_to_clean = os.path.join(output_dir, 'all_recipes_combined_processed2.csv')

try:
    df_final_clean = pd.read_csv(input_file_to_clean)
    print(f"Fichier '{input_file_to_clean}' chargé avec succès. Taille initiale : {len(df_final_clean)} recettes.")
except FileNotFoundError:
    print(f"Erreur : Le fichier '{input_file_to_clean}' n'a pas été trouvé.")
    print("Assure-toi que tu as le bon chemin et nom de fichier.")
    exit()

# --- Nettoyage des colonnes : S'assurer qu'elles sont des chaînes et gérer les NaN/vides ---
# 1. Remplacer tous les NaN par des chaînes vides pour faciliter la vérification de longueur
df_final_clean['ingredients_text'] = df_final_clean['ingredients_text'].fillna('')
df_final_clean['steps_text'] = df_final_clean['steps_text'].fillna('')

# 2. Convertir explicitement en type chaîne pour toute la colonne, au cas où il y aurait d'autres types inattendus
df_final_clean['ingredients_text'] = df_final_clean['ingredients_text'].astype(str)
df_final_clean['steps_text'] = df_final_clean['steps_text'].astype(str)

print("\nColonnes 'ingredients_text' et 'steps_text' préparées (NaN remplacés par '').")


# --- Filtrage : Supprimer les lignes où ingredients_text OU steps_text est vide ---
initial_rows = len(df_final_clean)

# Condition: filtrer les lignes où la longueur de ingredients_text est 0 OU la longueur de steps_text est 0
# Nous nous basons sur la longueur après s'être assurés qu'elles sont des chaînes.
df_final_clean = df_final_clean[
    (df_final_clean['ingredients_text'].str.strip() != '') &
    (df_final_clean['steps_text'].str.strip() != '')
].copy() # .copy() pour éviter les avertissements

print(f"\nRecettes après suppression des lignes sans ingrédients ou étapes : {len(df_final_clean)} (supprimées: {initial_rows - len(df_final_clean)})")
print("Aperçu du DataFrame après le nettoyage final :")
display(df_final_clean.head())


```

```python
# --- Sauvegarde du Dataset Nettoyé Final ---
output_final_cleaned_path = os.path.join(output_dir, 'all_recipes_combined_processed3.csv')
df_final_clean.to_csv(output_final_cleaned_path, index=False)

print(f"\nDataset nettoyé final sauvegardé sous : {output_final_cleaned_path}")
print(f"Taille finale du dataset : {len(df_final_clean)} recettes.")
```
